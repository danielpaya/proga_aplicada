## Ejercicio 1. Mi primer objeto: la clase y el constructor

### 📘 Concepto

**Objetivo:** crear una clase con constructor y un método que muestre su información.

**Analogía:** una clase es como la ficha en blanco de un casillero. El **constructor** (`__init__`) es el momento en que alguien llena esa ficha con los datos. Cada dato guardado en la ficha es un **atributo** (`self.algo`), y lo que el objeto sabe hacer es un **método**.

**Requerimiento del proyecto:** RF-01 y RF-02.

### 🔍 Ejemplo documentado: un casillero del gimnasio

````python
# Definimos la clase Casillero: es el MOLDE para crear casilleros
class Casillero:

    # El constructor se ejecuta automáticamente cuando creamos un casillero
    def __init__(self, numero, tamano):
        self.numero = numero      # atributo: número del casillero (ej: 7)
        self.tamano = tamano      # atributo: tamaño ("Pequeño" o "Grande")
        self.ocupado = False      # atributo: todo casillero nuevo empieza libre

    # Método de servicio: muestra la información del casillero
    def mostrar_info(self):
        if self.ocupado:              # si el casillero está ocupado...
            estado = "Ocupado"
        else:                         # si no...
            estado = "Libre"
        print("Casillero", self.numero, "| Tamaño:", self.tamano, "| Estado:", estado)


# Programa principal: creamos un OBJETO a partir del molde
c1 = Casillero(7, "Grande")   # se llama al constructor con numero=7 y tamano="Grande"
c1.mostrar_info()             # le pedimos al objeto que se muestre
````

## Ejercicio 3. Prestar con candado de seguridad

### 📘 Concepto

**Objetivo:** crear un método que **cambie** un atributo, pero solo si se cumple una condición.

**Analogía:** un semáforo. Si está en verde (libre), el carro pasa y el semáforo cambia a rojo. Si ya está en rojo, no deja pasar a nadie.

**Requerimiento del proyecto:** RF-03 y RF-04 (el sistema no debe prestar una bicicleta que ya está prestada).

### 🔍 Ejemplo documentado: asignar un casillero solo si está libre

````python
class Casillero:
    def __init__(self, numero, tamano):
        self.numero = numero
        self.tamano = tamano
        self.ocupado = False

    def mostrar_info(self):
        if self.ocupado:
            estado = "Ocupado"
        else:
            estado = "Libre"
        print("Casillero", self.numero, "| Tamaño:", self.tamano, "| Estado:", estado)

    # Nuevo método: asigna el casillero SOLO si está libre
    def asignar(self):
        if not self.ocupado:                 # "not" invierte: si NO está ocupado...
            self.ocupado = True              # ...lo marcamos como ocupado
            print("Casillero", self.numero, "asignado. ¡Disfruta tu entrenamiento!")
        else:                                # si ya estaba ocupado...
            print("Lo sentimos: el casillero", self.numero, "ya está ocupado.")


c1 = Casillero(7, "Grande")
c1.asignar()        # primera vez: está libre, se asigna
c1.asignar()        # segunda vez: ya está ocupado, sale el aviso y NO cambia nada
c1.mostrar_info()
````

## Ejercicio 4. El camino de regreso: devolver

### 📘 Concepto

**Objetivo:** crear el método contrario al anterior, también con su validación.

**Analogía:** es el mismo semáforo, pero al revés: solo puede volver a verde si estaba en rojo.

**Requerimiento del proyecto:** RF-05.

### 🔍 Ejemplo documentado: liberar un casillero

````python
class Casillero:
    def __init__(self, numero, tamano):
        self.numero = numero
        self.tamano = tamano
        self.ocupado = False

    def mostrar_info(self):
        if self.ocupado:
            estado = "Ocupado"
        else:
            estado = "Libre"
        print("Casillero", self.numero, "| Tamaño:", self.tamano, "| Estado:", estado)

    def asignar(self):
        if not self.ocupado:
            self.ocupado = True
            print("Casillero", self.numero, "asignado. ¡Disfruta tu entrenamiento!")
        else:
            print("Lo sentimos: el casillero", self.numero, "ya está ocupado.")

    # Nuevo método: libera el casillero SOLO si estaba ocupado
    def liberar(self):
        if self.ocupado:                     # solo tiene sentido liberar uno ocupado
            self.ocupado = False
            print("Casillero", self.numero, "liberado. ¡Gracias!")
        else:
            print("Aviso: el casillero", self.numero, "ya estaba libre.")


c1 = Casillero(7, "Grande")
c1.asignar()
c1.liberar()        # estaba ocupado: se libera
c1.liberar()        # ya estaba libre: sale el aviso
c1.mostrar_info()
````

## Ejercicio 5. Guardar objetos en una lista

### 📘 Concepto

**Objetivo:** guardar varios objetos en una lista y recorrerla con `for`.

**Analogía:** la lista es una fila de casilleros. En cada posición hay un objeto, y con un `for` pasas por cada uno, de izquierda a derecha, como quien revisa la fila entera.

**Requerimiento del proyecto:** RF-01 y RF-02.

### 🔍 Ejemplo documentado: la fila de casilleros del gimnasio

````python
class Casillero:
    def __init__(self, numero, tamano):
        self.numero = numero
        self.tamano = tamano
        self.ocupado = False

    def mostrar_info(self):
        if self.ocupado:
            estado = "Ocupado"
        else:
            estado = "Libre"
        print("Casillero", self.numero, "| Tamaño:", self.tamano, "| Estado:", estado)

    def asignar(self):
        if not self.ocupado:
            self.ocupado = True
            print("Casillero", self.numero, "asignado. ¡Disfruta tu entrenamiento!")
        else:
            print("Lo sentimos: el casillero", self.numero, "ya está ocupado.")

    def liberar(self):
        if self.ocupado:
            self.ocupado = False
            print("Casillero", self.numero, "liberado. ¡Gracias!")
        else:
            print("Aviso: el casillero", self.numero, "ya estaba libre.")


# Una lista que guarda OBJETOS (no números ni textos sueltos)
casilleros = [Casillero(1, "Pequeño"),
              Casillero(2, "Grande"),
              Casillero(3, "Pequeño"),
              Casillero(4, "Grande")]

# El for toma cada objeto de la lista, uno por uno, y lo llama "cas"
for cas in casilleros:
    cas.mostrar_info()        # a cada objeto le pedimos que se muestre

# También podemos usar un objeto por su posición (la primera posición es 0)
print("El primer casillero de la fila es el número", casilleros[0].numero)
````

## Ejercicio 6. Contar según el estado

### 📘 Concepto

**Objetivo:** recorrer una lista de objetos y contar cuántos cumplen una condición.

**Analogía:** como contar sombrillas en la entrada de un edificio: una pila para las que están y otra para las que se llevaron. Cada vez que revisas una, sumas 1 en la pila que corresponde.

**Requerimiento del proyecto:** RF-09 (totales de prestadas y disponibles) y etapa 8 de DevSecOps (Monitorear).

### 🔍 Ejemplo documentado: ¿cuántos casilleros grandes están libres?

````python
# (Usamos la clase Casillero del ejercicio anterior; ejecútalo antes si hace falta)

casilleros = [Casillero(1, "Pequeño"),
              Casillero(2, "Grande"),
              Casillero(3, "Pequeño"),
              Casillero(4, "Grande"),
              Casillero(5, "Grande")]

casilleros[1].asignar()       # ocupamos el casillero de la posición 1 (el número 2)

grandes_libres = 0            # contador: empieza en cero
pequenos = 0                  # otro contador

for cas in casilleros:
    if cas.tamano == "Grande" and not cas.ocupado:   # dos condiciones al tiempo
        grandes_libres = grandes_libres + 1          # sumamos 1 a este contador
    if cas.tamano == "Pequeño":
        pequenos = pequenos + 1

print("Casilleros grandes libres:", grandes_libres)
print("Casilleros pequeños en total:", pequenos)
````

## Ejercicio 7. Buscar por código

### 📘 Concepto

**Objetivo:** buscar un objeto en la lista comparando uno de sus atributos, y avisar si no existe.

**Analogía:** el cartero revisa buzón por buzón hasta encontrar el número de la dirección. Si recorre toda la calle y no lo encuentra, devuelve la carta. Para recordar si lo encontró o no, usa una **bandera**: una variable que empieza en `False` y cambia a `True` cuando lo halla.

**Requerimiento del proyecto:** RF-07.

### 🔍 Ejemplo documentado: buscar un casillero por su número

````python
casilleros = [Casillero(1, "Pequeño"),
              Casillero(2, "Grande"),
              Casillero(3, "Pequeño")]

numero_buscado = 2            # prueba cambiando este número por 9
encontrado = False            # la bandera empieza "abajo": aún no lo encontramos

for cas in casilleros:
    if cas.numero == numero_buscado:   # comparamos el ATRIBUTO con lo que buscamos
        cas.mostrar_info()
        encontrado = True              # subimos la bandera

# Solo DESPUÉS de revisar toda la lista sabemos si no estaba
if not encontrado:
    print("No existe el casillero número", numero_buscado)
````

## Ejercicio 8. Seguridad: validar al estudiante

### 📘 Concepto

**Objetivo:** crear una segunda clase con un método que **valide** un dato y devuelva `True` o `False` con `return`.

**Analogía:** el portero de un edificio revisa tu carné antes de dejarte pasar. No te abre la puerta: solo responde «sí es válido» o «no es válido».

**Dos herramientas nuevas y cortas:**
- `texto.isdigit()` responde `True` si el texto tiene **solo números**.
- `len(texto)` dice **cuántos caracteres** tiene.

**Requerimiento del proyecto:** RF-08 y RNF-02. Es la práctica **Sec** de DevSecOps.

### 🔍 Ejemplo documentado: validar la clave de un casillero

````python
class Usuario:
    def __init__(self, nombre, clave):
        self.nombre = nombre
        self.clave = clave            # la clave es un TEXTO, por ejemplo "4821"

    # Método que valida: devuelve True o False, no imprime nada
    def clave_valida(self):
        if self.clave.isdigit() and len(self.clave) == 4:   # solo números Y 4 caracteres
            return True
        else:
            return False


# Probamos con varias claves
u1 = Usuario("Sofía", "4821")
u2 = Usuario("Mateo", "48A1")         # tiene una letra
u3 = Usuario("Valeria", "12")         # es muy corta

print("Clave de Sofía válida:", u1.clave_valida())
print("Clave de Mateo válida:", u2.clave_valida())
print("Clave de Valeria válida:", u3.clave_valida())

# Usamos el resultado en una decisión
if u1.clave_valida():
    print("Bienvenida,", u1.nombre)
````

## Ejercicio 9. El mapa de la estación: lista 2D de objetos

### 📘 Concepto

**Objetivo:** organizar objetos en una **lista de listas** (filas y columnas) y recorrerla con dos `for`.

**Analogía:** un tablero de ajedrez. Cada casilla se ubica con fila y columna, y en cada casilla hay un objeto. `tablero[1][3]` significa «fila 1, columna 3» (recuerda que se empieza a contar en 0).

**Requerimiento del proyecto:** RF-10.

### 🔍 Ejemplo documentado: un pasillo de casilleros de 2 filas × 5 columnas

````python
pasillo = []                  # lista vacía: aquí van las filas
numero = 1                    # contador para numerar los casilleros

for fila in range(2):                 # 2 filas: fila 0 y fila 1
    fila_casilleros = []              # cada fila es una lista nueva
    for col in range(5):              # 5 casilleros por fila
        fila_casilleros.append(Casillero(numero, "Pequeño"))   # agregamos un objeto
        numero = numero + 1
    pasillo.append(fila_casilleros)   # agregamos la fila completa al pasillo

# Ocupamos dos casilleros usando [fila][columna]
pasillo[0][1].asignar()       # fila 0, columna 1  -> casillero 2
pasillo[1][4].asignar()       # fila 1, columna 4  -> casillero 10

# Dibujamos el mapa: L = libre, X = ocupado
print("Mapa del pasillo (L = libre, X = ocupado)")
for fila in range(2):
    linea = "Fila " + str(fila + 1) + ": "
    for col in range(5):
        if pasillo[fila][col].ocupado:
            linea = linea + "X "
        else:
            linea = linea + "L "
    print(linea)
````

## Ejercicio 10. Todo junto: el programa con menú

### 📘 Concepto

**Objetivo:** unir todas las piezas en un programa con **menú** que se repite con `while` hasta que el usuario decida salir.

**Analogía:** el menú es el control remoto del sistema: cada botón hace algo distinto, y el televisor sigue encendido hasta que presionas «apagar». Si alguien presiona un botón que no existe, el televisor no se daña: simplemente no hace nada y te avisa.

**Requerimiento del proyecto:** RF-11, RNF-01 y RNF-03.

### 🔍 Ejemplo documentado: el menú del gimnasio

````python
casilleros = [Casillero(1, "Pequeño"), Casillero(2, "Grande"), Casillero(3, "Pequeño")]

opcion = ""                           # empieza vacía para que el while arranque
while opcion != "4":                  # se repite MIENTRAS no elijan salir
    print("")
    print("===== Casilleros del gimnasio =====")
    print("1. Ver casilleros")
    print("2. Asignar un casillero")
    print("3. Liberar un casillero")
    print("4. Salir")
    opcion = input("Elige una opción: ")

    if opcion == "1":
        for cas in casilleros:
            cas.mostrar_info()

    elif opcion == "2":
        texto = input("Número del casillero: ")
        if texto.isdigit():                        # validamos ANTES de convertir
            numero = int(texto)                    # int() convierte el texto a número
            encontrado = False
            for cas in casilleros:
                if cas.numero == numero:
                    cas.asignar()
                    encontrado = True
            if not encontrado:
                print("Ese casillero no existe.")
        else:
            print("Debes escribir solo números.")

    elif opcion == "3":
        texto = input("Número del casillero: ")
        if texto.isdigit():
            numero = int(texto)
            encontrado = False
            for cas in casilleros:
                if cas.numero == numero:
                    cas.liberar()
                    encontrado = True
            if not encontrado:
                print("Ese casillero no existe.")
        else:
            print("Debes escribir solo números.")

    elif opcion == "4":
        print("¡Hasta pronto!")

    else:                                          # cualquier otra cosa
        print("Opción no válida. Escribe un número del 1 al 4.")
````
